```python
from hmmlearn import hmm
import numpy as np
```

```python
obs_names = {'r': 2, 'g': 1, 'b': 2}
stari = ['camera1', 'camera2', 'camera3', 'camera4']

model=hmm.CategoricalHMM(n_components=4)
model.startprob_=np.array([0.2, 0.2, 0.2, 0.2])
model.transmat_=np.array([0.2, 0.4, 0, 0.4], [0.4, 0.2, 0.4, 0], [0, 0.4, 0.2, 0.4], [0.4, 0, 0.4, 0.2])
model.emissionprob_=np.array([0.7, 0.2, 0.1], [0.2, 0.6, 0.2], [0.1, 0.3, 0.6], [0.5, 0.2, 0.3])

obs_seq=np.array ([[ obs_names ['r']], [ obs_names ['g']], [ obs_names ['b']]])
log_prob, secv_camere = model.decode(obs_seq , algorithm ="viterbi")
secv = [stari[i] for i in secv_camere]

print (f"Secventa de camere: {secv}")
print (f"Probabilitatea comuna: {np.exp(log_prob):.4f}")
```

```python
import numpy as np
import pyAgrum as gum
import pyAgrum.lib.notebook as gnb
```

```python
mn = gum.MarkovNet()

a = mn.add(gum.LabelizedVariable("A", "A", 2))
b = mn.add(gum.LabelizedVariable("B", "B", 2))
c = mn.add(gum.LabelizedVariable("C", "C", 2))
d = mn.add(gum.LabelizedVariable("D", "D", 2))
e = mn.add(gum.LabelizedVariable("E", "E", 2))

mn.addEdges ([(a, b), (b ,c), (c ,d), (d ,e)])
```

```python
import pymc as pm
import numpy as np
import arviz as az
import matplotlib .pyplot as plt
from scipy import stats
```

```python
alpha_prior , beta_prior = 2,8
N=20
y=5

a_post = alpha_prior + y
b_post = beta_prior + (N - y)

media_post = a_post / (a_post + b_post)

print(f"Posterior: Beta({a_post}, {b_post})")
print(f"Media a posteriori: {media_post:.4f}")

esantioane = stats.beta.rvs(a_post , b_post , size =20000 , random_state =42)

az. plot_posterior (esantioane , hdi_prob =0.94)
plt.show ()
```
