```python
import numpy as np
N = 100000
red = 0
for _ in range(N):
    urna = ['R', 'R', 'R', 'A', 'A', 'A', 'A', 'N', 'N']
    zar = np.random.choice([1, 2, 3, 4, 5, 6])

    if zar in [2, 3, 5]:
        urna.append('N')
    elif zar == 6:
        urna.append('R')
    else:
        urna.append('A')

    bila_extrasa = np.random.choice(urna)
    if bila_extrasa == 'R':
        red += 1

print(f"Probabilitatea estimată prin simulare: {red / N:.4f}")
print(f"Probabilitatea teoretică este: 19/60 = {19/60:.4f}")
```
