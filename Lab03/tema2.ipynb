```python
import numpy as np
import pandas as pd
```

**Exercitiul 1**

Extensie Python

```python
N = 100000
X = np.random.randint(1, 7, N)
Y = np.random.randint(1, 7, N)
suma = X + Y

x_par = (X % 2 == 0)
prob_1 = np.sum((suma >= 10) & x_par) / np.sum(x_par)
print(f"1. Probabilitate estimată: {prob_1:.4f} | Exactă: {2/9:.4f}")

cel_putin_unul_par = (X % 2 == 0) | (Y % 2 == 0)
prob_2 = np.sum((suma >= 10) & cel_putin_unul_par) / np.sum(cel_putin_unul_par)
print(f"2. Probabilitate estimată: {prob_2:.4f} | Exactă: {5/27:.4f}")

ambele_pare = (X % 2 == 0) & (Y % 2 == 0)
prob_3 = np.sum(ambele_pare & (suma == 8)) / np.sum(suma == 8)
print(f"3. Probabilitate estimată: {prob_3:.4f} | Exactă: {3/5:.4f}")
```

**Exercitiul 3**

Extensie Python

```python
nume_fisier = 'simpson.csv'
df = pd.read_csv(nume_fisier)

df['Rata(%)'] = (df['Recenzii_Pozitive'] / df['Total_Clienți']) * 100

print("1. Analiza pe Subgrupuri\n")
print(df[['Restaurant', 'Tip_Client', 'Rata(%)']])

df_global = df.groupby('Restaurant')[['Recenzii_Pozitive', 'Total_Clienți']].sum().reset_index()
df_global['Rata_Globală(%)'] = (df_global['Recenzii_Pozitive'] / df_global['Total_Clienți']) * 100

print("\n2. Analiza Globală Agregată\n")
print(df_global[['Restaurant', 'Rata_Globală(%)']])
```
